```python
# Initialize account database with Account Numbers, User Names, PINs, and Balances
accounts = {
    "101": {
        "user_name": "John Doe",
        "pin": "1234",
        "savings": 10000.0,
        "current": 5000.0
    },
    "102": {
        "user_name": "Jane Smith",
        "pin": "4321",
        "savings": 20000.0,
        "current": 15000.0
    }
}

# Step 1: Prompt the user to enter their Account Number
acc_no = input("Enter Account Number: ")

# Check if the entered Account Number exists in the system database
if acc_no in accounts:
    # Retrieve the specific account details dictionary for this user
    user = accounts[acc_no]

    # Step 2: Show the main options after entering the Account Number
    print("\n--- Options ---")
    print("1. Balance Enquiry")
    print("2. A/c User Name")
    print("3. Withdrawal")

    # Prompt user to select an option from the menu
    option = input("Choose an option (1-3): ")

    # Handle Option 1: Balance Enquiry
    if option == "1":
        # Display the Savings Account balance
        print("Savings Balance: ₹" + str(user["savings"]))
        # Display the Current Account balance
        print("Current Balance: ₹" + str(user["current"]))

    # Handle Option 2: A/c User Name
    elif option == "2":
        # Display the Account User Name
        print("Account User Name:", user["user_name"])

    # Handle Option 3: Withdrawal
    elif option == "3":
        # Prompt user to select account type (1 for Savings / 2 for Current)
        acc_type = input("Choose Account Type (1 for Savings / 2 for Current): ")

        # Check if user selected Savings Account (Option 1)
        if acc_type == "1":
            # Set key string to 'savings'
            bal_key = "savings"
        # Check if user selected Current Account (Option 2)
        elif acc_type == "2":
            # Set key string to 'current'
            bal_key = "current"
        # Handle invalid account type choice
        else:
            # Display invalid account type message
            print("Invalid account type selected.")
            # Set bal_key to None to indicate invalid selection
            bal_key = None

        # Proceed only if a valid account type was chosen
        if bal_key is not None:
            # Prompt user to enter the amount to withdraw
            amount = float(input("Enter the amount to withdraw: "))

            # Check if the entered amount is greater than the account balance
            if amount > user[bal_key]:
                # Ask the user to enter the correct amount if balance is insufficient
                print("Amount exceeds account balance! Please run the program again and enter a correct amount.")
            # Check if amount is less than or equal to zero
            elif amount <= 0:
                # Ask user to enter a valid positive amount
                print("Invalid amount! Please run the program again and enter a correct positive amount.")
            # If amount is valid and within available balance
            else:
                # Prompt user to enter their PIN after entering the amount
                entered_pin = input("Enter PIN: ")

                # Check if the entered PIN matches the correct PIN
                if entered_pin == user["pin"]:
                    # Deduct entered amount from the account balance
                    user[bal_key] -= amount
                    # Update and replace the result balance in the respective balance position
                    # Show withdrawal success message
                    print("\nWithdrawal Successful!")
                    # Show updated balance
                    print("Remaining Balance: ₹" + str(user[bal_key]))
                    # Display "thanks for visiting again" message
                    print("Thanks for visiting again!")
                # Handle incorrect PIN entry
                else:
                    # Display incorrect PIN message and end program
                    print("Incorrect PIN! Access denied.")

    # Handle invalid main menu selection
    else:
        # Display invalid option message
        print("Invalid option selected.")

# Handle scenario where Account Number is invalid
else:
    # Display invalid Account Number message
    print("Account Number not found!")
```
