```python
accounts = {
    "101": {
        "user_name": "John Doe",
        "pin": "1234",
        "savings": 10000.0,
        "current": 5000.0
    },
    "102": {
        "user_name": "Jane Smith",
        "pin": "4321",
        "savings": 20000.0,
        "current": 15000.0
    }
}
acc_no = input("Enter Account Number: ")
if acc_no in accounts:
    user = accounts[acc_no]
    print("\n--- Options ---")
    print("1. Balance Enquiry")
    print("2. A/c User Name")
    print("3. Withdrawal")
    option = input("Choose an option (1-3): ")
    if option == "1":
        print("Savings Balance: ₹" + str(user["savings"]))
        print("Current Balance: ₹" + str(user["current"]))
    elif option == "2":
        print("Account User Name:", user["user_name"])
    elif option == "3":
        # Prompt user to select account type (1 for Savings / 2 for Current)
        acc_type = input("Choose Account Type (1 for Savings / 2 for Current): ")
        if acc_type == "1":
            bal_key = "savings"
        elif acc_type == "2":
            bal_key = "current"
        else:
            print("Invalid account type selected.")
            bal_key = None
        if bal_key is not None:
            amount = float(input("Enter the amount to withdraw: "))
            if amount > user[bal_key]:
                print("Amount exceeds account balance! Please run the program again and enter a correct amount.")
            elif amount <= 0:
                print("Invalid amount! Please run the program again and enter a correct positive amount.")
            else:
                entered_pin = input("Enter PIN: ")
                if entered_pin == user["pin"]:
                    user[bal_key] -= amount
                    print("\nWithdrawal Successful!")
                    print("Remaining Balance: ₹" + str(user[bal_key]))
                    print("Thanks for visiting again!")
                else:
                    print("Incorrect PIN! Access denied.")
    else:
        print("Invalid option selected.")
else:
    print("Account Number not found!")
```
