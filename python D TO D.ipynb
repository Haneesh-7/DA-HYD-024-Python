```python
class Solution:
    def armstrongNumber (self, n):
        original = n
        total = 0

        while n > 0:
            digit = n % 10
            total = total + digit ** 3
            n = n // 10

        if total == original:
            return True
        else:
            return False
```

```python
sol = Solution()
num = int(input("Enter a number to check: "))
print(f"Is {num} an Armstrong number? {sol.armstrongNumber(num)}")
```

```python
class Solution:
  def checkString(self, s):
    v = 0
    c = 0
    vowels = set('aeiou')

    for ch in s:
        if ch in vowels:
            v += 1
        else:
            c += 1

    if v > c:
        print("Yes")
    elif v < c:
        print("No")
    else:
        print("Same")
```

```python
sol = Solution()
s = "education"
sol.checkString(s)
```

```python
class Solution:
    def areAnagrams(self, s1, s2):
        return sorted(s1) == sorted(s2)
```

```python
sol = Solution()
str1 = "listen"
str2 = "silent"
result = sol.areAnagrams(str1, str2)
print(f"Are '{str1}' and '{str2}' anagrams? {result}")
```

```python
class Solution:
    def fibonacciNumbers(self, n):
        result = []
        a = 0
        b = 1
        for i in range(n+1):
            result.append(a)
            a, b = b, a + b
        return result
```

```python
sol=Solution()
num = int(input("Enter a number to check: "))
print(f"Is {num} an fibonacciNumbers? {sol.fibonacciNumbers(num)}")
```

```python
class Solution:
    def sumOfDigits(self, n):
        total = 0
        p = 1
        while p <= n:
            higher = n // (p * 10)
            current = (n // p) % 10
            lower = n % p
            total += higher * 45 * p
            total += (current * (current - 1) // 2) * p
            total += current * (lower + 1)
            p *= 10
        return total
```

```python
sol=Solution()
num = int(input("Enter a number: "))
ans = sol.sumOfDigits(num)
print(f"Total digit sum up to {num} is: {ans}")
```

```python
class Solution:
    def commonElements(self, a, b):
        freq = {}
        result = []

        for x in a:
            freq[x] = freq.get(x, 0) + 1

        for x in b:
            if x in freq and freq[x] > 0:
                result.append(x)
                freq[x] -= 1

        result.sort()
        return result
```

```python
sol = Solution()
a = [3, 4, 2, 2, 4]
b = [3, 2, 2, 7]
output = sol.commonElements(a, b)
print(f"Common elements: {output}")
```

```python
class Solution:
    def getSecondLargest(self, arr):
        largest = -1
        second = -1

        for i in arr:
            if i > largest:
                second = largest
                largest = i
            elif i > second and i != largest:
                second = i

        return second
```

```python
sol = Solution()
arr = [12, 35, 1, 10, 34, 1]
result = sol.getSecondLargest(arr)
print(f"Array: {arr}")
print(f"Second largest element: {result}")
```

```python
class Solution:
    def maxSubarraySum(self, arr, k):
        window_sum = 0

        # First window
        for i in range(k):
            window_sum += arr[i]

        max_sum = window_sum

        # Slide the window
        for i in range(k, len(arr)):
            window_sum = window_sum + arr[i] - arr[i-k]

            if window_sum > max_sum:
                max_sum = window_sum

        return max_sum
```

```python
sol = Solution()
arr = [100, 200, 300, 400]
k = 2
result = sol.maxSubarraySum(arr, k)
print(f"Array: {arr}")
print(f"Maximum sum of subarray of size {k}: {result}")
```
