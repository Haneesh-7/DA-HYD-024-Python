```python
n = int(input("Enter a number: "))
original = n
reverse = 0
while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10
if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
n = int(input("Enter a number: "))
original = n
count = 0
temp = n
while temp > 0:
    count = count + 1
    temp = temp // 10
total = 0
temp = n
while temp > 0:
    digit = temp % 10
    total = total + digit ** count
    temp = temp // 10
if total == original:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

```python
n = int(input("Enter a number: "))
total = 0
i = 1
while i < n:
    if n % i == 0:
        total = total + i
    i = i + 1
if total == n:
    print("Perfect Number")
else:
    print("Not Perfect Number")
```

```python
n = int(input("Enter a number: "))
count = 0
i = 1
while i <= n:
    if n % i == 0:
        count = count + 1
    i = i + 1
if count == 2:
    print("Prime Number")
else:
    print("Not Prime Number")
```

```python
n = int(input("Enter a number: "))
original = n
total = 0
while n > 0:
    digit = n % 10
    factorial = 1
    i = 1
    while i <= digit:
        factorial = factorial * i
        i = i + 1
    total = total + factorial
    n = n // 10
if total == original:
    print("Strong Number")
else:
    print("Not Strong Number")
```
