```python
salary = float(input("Enter Salary: "))
experience = float(input("Enter Years of Experience: "))
if experience < 1:
    bonus_percentage = 0
elif experience == 1:
    bonus_percentage = 0.05
elif 2 <= experience <= 5:
    bonus_percentage = 0.10
else:
    bonus_percentage = 0.20
bonus = salary * bonus_percentage
final_salary = salary + bonus
print("Bonus Amount =", bonus)
print("Final Salary =", final_salary)
```

```python
math = float(input("Enter Math marks: "))
physics = float(input("Enter Physics marks: "))
chemistry = float(input("Enter Chemistry marks: "))
average = (math + physics + chemistry) / 3
if math >= 60 and physics >= 50 and chemistry >= 50 and average >= 60:
    print("Eligible")
else:
    print("Not Eligible")
```

```python
purchase_amount = float(input("Enter Purchase Amount: "))
days_delayed = int(input("Enter Days Delayed: "))
if purchase_amount >= 5000:
    discount_rate = 0.20
elif purchase_amount >= 2000:
    discount_rate = 0.10
else:
    discount_rate = 0.0
discount = purchase_amount * discount_rate
amount_after_discount = purchase_amount - discount
if days_delayed == 0:
    penalty = 0
elif 1 <= days_delayed <= 5:
    penalty = 100
else:
    penalty = 300
final_amount = amount_after_discount + penalty
print("Discount: ₹" + str(int(discount)))
print("Penalty: ₹" + str(penalty))
print("Final Amount: ₹" + str(int(final_amount)))
```

```python
total_sum = 0
while True:
    number = int(input("Enter a number (0 to stop): "))
    if number == 0:
        break
    total_sum += number
print("Total sum:", total_sum)
```

```python
count = 0
while True:
    number = int(input("Enter a number (-1 to stop): "))
    if number == -1:
        break
    count += 1
print("Total numbers entered:", count)
```

```python
total_sum = 0
while True:
    number = int(input("Enter a number (negative number to stop): "))
    if number < 0:
        break
    total_sum += number
print("Total sum:", total_sum)
```

```python
even_count = 0
while True:
    number = int(input("Enter a number (0 to stop): "))
    if number == 0:
        break
    if number % 2 == 0:
        even_count += 1
print("Total even numbers entered:", even_count)
```
